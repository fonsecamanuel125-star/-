```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 创建DataFrame
df = pd.DataFrame({
    "MonthlyCharges": [10, 20, 30, 40, 50],
    "TotalCharges": [100, 200, 300, 400, 500],
    "Label": ["A", "B", "C", "D", "E"]
})

# 有一个'tenure'列，用MonthlyCharges代替
df['tenure'] = df['MonthlyCharges']

# 排序，按照'Label'列排序
df = df.sort_values(by="Label")

# 切片操作，取第二行到第四行
cut_df = df.iloc[1:4]

# 使用seaborn的catplot绘制分类计数图
sns.catplot(x="tenure", kind="count", hue="Label", palette='pastel', data=cut_df)

# 显示图表
plt.show()
```
